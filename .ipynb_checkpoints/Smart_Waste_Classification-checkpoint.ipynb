```python
#Import libraries
import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

from tensorflow.keras import layers, models
```

```python
#Check TensorFlow version
print("TensorFlow version:", tf.__version__)
```

```python
#Set dataset path
dataset_path = r"D:\waste_classificatiob\Garbage_dataset\original"
```

```python
#Define the 10 classes
class_names = [
    "battery",
    "biological",
    "cardboard",
    "clothes",
    "glass",
    "metal",
    "paper",
    "plastic",
    "shoes",
    "trash"
]

print("Number of classes:", len(class_names))
print("Classes:", class_names)
```

```python
#Check folders
for class_name in class_names:
    folder_path = os.path.join(dataset_path, class_name)

    if os.path.exists(folder_path):
        print("Found:", class_name)
    else:
        print("Missing:", class_name)
```

```python
#Count images in every class
image_extensions = (".jpg", ".jpeg", ".png", ".bmp", ".webp")

class_counts = {}

for class_name in class_names:
    folder_path = os.path.join(dataset_path, class_name)

    count = 0

    for file_name in os.listdir(folder_path):
        if file_name.lower().endswith(image_extensions):
            count += 1

    class_counts[class_name] = count

class_counts
```

```python
#Display class distribution
plt.figure(figsize=(12, 6))

plt.bar(class_counts.keys(), class_counts.values())

plt.title("Waste Dataset Class Distribution")
plt.xlabel("Waste Category")
plt.ylabel("Number of Images")

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#Display number of images
total_images = sum(class_counts.values())

print("Total number of images:", total_images)

for class_name, count in class_counts.items():
    print(f"{class_name}: {count}")
```

```python
#Find all image paths
image_paths = []
image_labels = []

for label, class_name in enumerate(class_names):

    folder_path = os.path.join(dataset_path, class_name)

    for file_name in os.listdir(folder_path):

        if file_name.lower().endswith(image_extensions):

            file_path = os.path.join(folder_path, file_name)

            image_paths.append(file_path)
            image_labels.append(label)

print("Total images found:", len(image_paths))
print("Total labels:", len(image_labels))
```

```python
#Create a DataFrame
data = pd.DataFrame({
    "image_path": image_paths,
    "label": image_labels
})

data["class_name"] = data["label"].apply(
    lambda x: class_names[x]
)

data.head()
```

```python
#Check the dataset
print("Dataset shape:", data.shape)

print("\nClass distribution:")
print(data["class_name"].value_counts())
```

```python
#Display sample images
plt.figure(figsize=(15, 10))

sample_data = data.sample(20, random_state=42)

for i, (_, row) in enumerate(sample_data.iterrows()):

    image = cv2.imread(row["image_path"])

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.subplot(4, 5, i + 1)

    plt.imshow(image)
    plt.title(row["class_name"])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Check one image before preprocessing
sample_path = image_paths[1]

original_image = cv2.imread(sample_path)

print("Image path:", sample_path)
print("Original image shape:", original_image.shape)
print("Minimum pixel value:", original_image.min())
print("Maximum pixel value:", original_image.max())
```

```python
#Image preprocessing
IMAGE_SIZE = 160

def preprocess_image(image_path):
    image = cv2.imread(image_path)

    if image is None:
        raise ValueError(f"Cannot read image: {image_path}")

    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    image = cv2.resize(image, (IMAGE_SIZE, IMAGE_SIZE))

    # Keep pixel values between 0 and 1
    image = image.astype(np.float32) / 255.0

    return image
```

```python
#Test preprocessing
processed_image = preprocess_image(sample_path)

print("Processed image shape:", processed_image.shape)

print("Minimum pixel value after normalization:", processed_image.min())
print("Maximum pixel value after normalization:", processed_image.max())
```

```python
#Show original and processed image
original = cv2.imread(sample_path)

original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)

processed = preprocess_image(sample_path)

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)

plt.imshow(original)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)

plt.imshow(processed)
plt.title("Preprocessed Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Split dataset
train_data, temp_data = train_test_split(
    data,
    test_size=0.30,
    random_state=42,
    stratify=data["label"]
)

validation_data, test_data = train_test_split(
    temp_data,
    test_size=0.50,
    random_state=42,
    stratify=temp_data["label"]
)

print("Training images:", len(train_data))
print("Validation images:", len(validation_data))
print("Testing images:", len(test_data))
```

```python
#Check class distribution after splitting
print("Training distribution:")
print(train_data["class_name"].value_counts())

print("\nValidation distribution:")
print(validation_data["class_name"].value_counts())

print("\nTesting distribution:")
print(test_data["class_name"].value_counts())
```

```python
#Create TensorFlow dataset
def load_and_preprocess(path, label):

    def process_image(path_string):

        path_string = path_string.decode("utf-8")

        image = preprocess_image(path_string)

        return image.astype(np.float32)

    image = tf.numpy_function(
        process_image,
        [path],
        tf.float32
    )

    image.set_shape((IMAGE_SIZE, IMAGE_SIZE, 3))

    return image, label
```

```python
#Function to create dataset
BATCH_SIZE = 32

def create_dataset(dataframe, shuffle=False):

    paths = dataframe["image_path"].values
    labels = dataframe["label"].values

    dataset = tf.data.Dataset.from_tensor_slices(
        (paths, labels)
    )

    dataset = dataset.map(
        load_and_preprocess,
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if shuffle:
        dataset = dataset.shuffle(
            buffer_size=1000,
            seed=42
        )

    dataset = dataset.batch(BATCH_SIZE)

    dataset = dataset.prefetch(
        tf.data.AUTOTUNE
    )

    return dataset
```

```python
#Create train, validation and test datasets
train_dataset = create_dataset(
    train_data,
    shuffle=True
)

validation_dataset = create_dataset(
    validation_data,
    shuffle=False
)

test_dataset = create_dataset(
    test_data,
    shuffle=False
)
```

```python
#Check one batch
for images, labels in train_dataset.take(1):

    print("Image batch shape:", images.shape)
    print("Label batch shape:", labels.shape)

    print("Minimum pixel value:",
          tf.reduce_min(images).numpy())

    print("Maximum pixel value:",
          tf.reduce_max(images).numpy())

    print("First 10 labels:",
          labels[:10].numpy())
```

```python
#Display preprocessed training images
plt.figure(figsize=(12, 8))

for images, labels in train_dataset.take(1):

    for i in range(12):

        plt.subplot(3, 4, i + 1)

        plt.imshow(images[i])

        plt.title(class_names[labels[i].numpy()])

        plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Calculate class weights
from sklearn.utils.class_weight import compute_class_weight

classes = np.unique(train_data["label"])

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=train_data["label"]
)

class_weights = dict(
    zip(classes, weights)
)

print("Class weights:")

for class_id, weight in class_weights.items():

    print(
        class_names[class_id],
        ":", 
        round(weight, 3)
    )
```

```python
# Transfer learning using MobileNetV2

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(IMAGE_SIZE, IMAGE_SIZE, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained layers 
base_model.trainable = False

model = models.Sequential([
    layers.Input(shape=(IMAGE_SIZE, IMAGE_SIZE, 3)),

    # Data augmentation
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.08),
    layers.RandomZoom(0.10),

    # MobileNetV2 expects values in [-1, 1]
    layers.Rescaling(scale=2.0, offset=-1.0),

    base_model,

    layers.GlobalAveragePooling2D(),

    layers.Dropout(0.35),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.30),

    layers.Dense(len(class_names), activation="softmax")
])

model.summary()
```

```python
#CELL 27 — Display CNN model
model.summary()
```

```python
#Compile model

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
#Train the model

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.2,
    patience=2,
    min_lr=1e-7,
    verbose=1
)

EPOCHS = 25

history = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=EPOCHS,
    class_weight=class_weights,
    callbacks=[early_stopping, reduce_lr]
) 
```

```python
#Fine-tuning

base_model.trainable = True

# Freeze most layers; fine-tuning only the last 30
for layer in base_model.layers[:-30]:
    layer.trainable = False

# Keep Batch Normalization layers frozen
for layer in base_model.layers:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history_fine = model.fit(
    train_dataset,
    validation_data=validation_dataset,
    epochs=10,
    class_weight=class_weights,
    callbacks=[early_stopping, reduce_lr]
)
```

```python
#Training accuracy graph
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Training and Validation Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.grid()

plt.show()
```

```python
#Training loss graph
plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training and Validation Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.grid()

plt.show()
```

```python
#Evaluate model on test data
test_loss, test_accuracy = model.evaluate(
    test_dataset
)

print("Test Loss:", test_loss)

print(
    "Test Accuracy:",
    round(test_accuracy * 100, 2),
    "%"
)
```

```python
#Get predictions
y_true = []
y_pred = []

for images, labels in test_dataset:

    predictions = model.predict(
        images,
        verbose=0
    )

    predicted_labels = np.argmax(
        predictions,
        axis=1
    )

    y_true.extend(
        labels.numpy()
    )

    y_pred.extend(
        predicted_labels
    )

y_true = np.array(y_true)
y_pred = np.array(y_pred)

print("Predictions completed.")
```

```python
#Calculate accuracy
accuracy = accuracy_score(
    y_true,
    y_pred
)

print(
    "Accuracy:",
    round(accuracy * 100, 2),
    "%"
)
```

```python
#Precision
precision = precision_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print(
    "Precision:",
    round(precision * 100, 2),
    "%"
)
```

```python
#Recall
recall = recall_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print(
    "Recall:",
    round(recall * 100, 2),
    "%"
)
```

```python
#F1-score
f1 = f1_score(
    y_true,
    y_pred,
    average="weighted",
    zero_division=0
)

print(
    "F1-score:",
    round(f1 * 100, 2),
    "%"
)
```

```python
#Display all evaluation results
evaluation_results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],

    "Score": [
        accuracy * 100,
        precision * 100,
        recall * 100,
        f1 * 100
    ]
})

evaluation_results["Score"] = evaluation_results["Score"].round(2)

evaluation_results
```

```python
#Classification report
print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        zero_division=0
    )
)
```

```python
#Confusion matrix
cm = confusion_matrix(
    y_true,
    y_pred
)

plt.figure(figsize=(10, 8))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=class_names,
    yticklabels=class_names
)

plt.title("Confusion Matrix")

plt.xlabel("Predicted Class")

plt.ylabel("Actual Class")

plt.xticks(rotation=45)

plt.yticks(rotation=0)

plt.tight_layout()

plt.show()
```

```python
#Create disposal recommendations
recommendations = {

    "battery":
        "Dispose of batteries at a battery or e-waste collection center.",

    "biological":
        "Send biological waste for composting or proper biological waste treatment.",

    "cardboard":
        "Send cardboard for recycling. Keep it dry and clean if possible.",

    "clothes":
        "Reuse or donate clothes. If they cannot be reused, send them for textile recycling.",

    "glass":
        "Send glass waste to a glass recycling facility.",

    "metal":
        "Send metal waste to a metal recycling facility.",

    "paper":
        "Send paper waste for paper recycling.",

    "plastic":
        "Send plastic waste to a plastic recycling facility.",

    "shoes":
        "Reuse or donate shoes. If they cannot be reused, use an appropriate recycling facility.",

    "trash":
        "Dispose of general trash through the appropriate waste disposal system."
}
```

```python
#Test unseen image
test_image_path = r"D:\waste_classificatiob\Garbage_dataset\my_test_images\battery.jpg"
```

```python
#Prediction function
def predict_waste(image_path):

    # Read original image
    original = cv2.imread(image_path)

    if original is None:
        print("Image could not be loaded.")
        return

    # Convert for displaying
    display_image = cv2.cvtColor(
        original,
        cv2.COLOR_BGR2RGB
    )

    # Preprocess image
    processed = preprocess_image(
        image_path
    )

    # Add batch dimension
    image_array = np.expand_dims(
        processed,
        axis=0
    )

    # Get prediction
    prediction = model.predict(
        image_array,
        verbose=0
    )[0]

    # Find class with highest probability
    predicted_index = np.argmax(
        prediction
    )

    predicted_class = class_names[
        predicted_index
    ]

    confidence = prediction[
        predicted_index
    ] * 100

    # Display image
    plt.figure(figsize=(6, 5))

    plt.imshow(display_image)

    plt.title(
        f"Prediction: {predicted_class}\n"
        f"Confidence: {confidence:.2f}%"
    )

    plt.axis("off")

    plt.show()

    # Print results
    print(
        "Predicted Waste Category:",
        predicted_class
    )

    print(
        f"Confidence: {confidence:.2f}%"
    )

    print("\nRecommendation:")

    print(
        recommendations[predicted_class]
    )
```

```python
#Run prediction
test_image_path = r"D:\waste_classificatiob\Garbage_dataset\my_test_images\battery.jpg"

predict_waste(test_image_path)
```

```python
#Show probability for all classes
def show_all_probabilities(image_path):

    processed = preprocess_image(
        image_path
    )

    image_array = np.expand_dims(
        processed,
        axis=0
    )

    prediction = model.predict(
        image_array,
        verbose=0
    )[0]

    probability_data = pd.DataFrame({
        "Waste Category": class_names,
        "Probability (%)": prediction * 100
    })

    probability_data = probability_data.sort_values(
        "Probability (%)",
        ascending=False
    )

    probability_data["Probability (%)"] = (
        probability_data["Probability (%)"]
        .round(2)
    )

    print(probability_data)

    plt.figure(figsize=(10, 5))

    plt.bar(
        probability_data["Waste Category"],
        probability_data["Probability (%)"]
    )

    plt.title(
        "Prediction Probability for Each Waste Category"
    )

    plt.xlabel("Waste Category")

    plt.ylabel("Probability (%)")

    plt.xticks(rotation=45)

    plt.tight_layout()

    plt.show()
```

```python
#Run probability analysis
show_all_probabilities(
    test_image_path
)
```

```python
#Create a simple final prediction function
def classify_waste(image_path):

    processed = preprocess_image(
        image_path
    )

    image_array = np.expand_dims(
        processed,
        axis=0
    )

    prediction = model.predict(
        image_array,
        verbose=0
    )[0]

    predicted_index = np.argmax(
        prediction
    )

    category = class_names[
        predicted_index
    ]

    confidence = prediction[
        predicted_index
    ] * 100

    print("=" * 50)

    print("SMART WASTE CLASSIFICATION SYSTEM")

    print("=" * 50)

    print(
        "Waste Category:",
        category
    )

    print(
        f"Confidence: {confidence:.2f}%"
    )

    print("\nRecommended Action:")

    print(
        recommendations[category]
    )

    print("=" * 50)
```

```python
#Final demonstration
classify_waste(test_image_path)
```
